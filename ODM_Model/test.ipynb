```python
import os

base = "accident_dataset"

print("=== Accident Dataset Structure ===\n")
for root, dirs, files in os.walk(base):
    dirs[:] = [d for d in dirs if d != "__pycache__"]
    level   = root.replace(base, "").count(os.sep)
    indent  = "  " * level
    print(f"{indent}{os.path.basename(root)}/")
    for f in files:
        size = os.path.getsize(os.path.join(root, f))
        print(f"{indent}  {f}  ({size/1024:.1f} KB)")
```

```python
import json

# Find the annotation file
for root, dirs, files in os.walk("accident_dataset"):
    for f in files:
        if f.endswith(".json"):
            path = os.path.join(root, f)
            with open(path, "r") as fp:
                data = json.load(fp)
            
            print(f"=== {path} ===\n")
            print(f"Images      : {len(data.get('images', []))}")
            print(f"Annotations : {len(data.get('annotations', []))}")
            print(f"Categories  :")
            for cat in data.get("categories", []):
                print(f"  id={cat['id']} name={cat['name']}")
```
